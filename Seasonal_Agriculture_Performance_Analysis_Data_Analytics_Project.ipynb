**Problem Statement**

Agricultural activities are influenced by seasonal variations in environmental conditions,
farming practices, resource availability and market conditions. As a result, agricultural
performance may differ from one season to another.
However, raw agricultural data does not clearly explain how agricultural performance changes
across seasons or what patterns can be observed in different seasonal conditions.

The problem is to analyze the given agricultural dataset and investigate seasonal differences
in agricultural performance by identifying meaningful patterns, trends, relationships and
variations within the available data.


**Project Objective**

To analyze agricultural data across different seasons, identify variations in farming performance, discover important trends and patterns, and generate meaningful insights that help in understanding seasonal agricultural outcomes.


To analyze agricultural performance across different seasons.

To identify seasonal patterns and changes in agricultural activities.

To compare agricultural outcomes between different seasons.

To examine relationships between important factors in the dataset.

To discover meaningful trends and variations using data analysis.

To present the findings through clear and informative visualizations.

To derive useful insights from the available agricultural data.


**Project Description**


Seasonal Agriculture Performance Analysis is a data analytics project that studies agricultural performance across different seasons and farming conditions. The project analyzes factors such as environmental conditions, farming practices, resource usage, crop production, and economic performance to identify meaningful patterns, trends, relationships, and differences. The insights obtained from the analysis help in understanding seasonal agricultural performance and support better agricultural planning.

```python
# Handle missing values

numeric_columns = df.select_dtypes(include='number').columns
categorical_columns = df.select_dtypes(include='object').columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Total missing values after handling:",
      df.isnull().sum().sum())
```

```python
# Check columns with missing values

missing = df.isnull().sum()

missing = missing[missing > 0]

if len(missing) == 0:
    print("No missing values found in the dataset.")
else:
    print("Columns with missing values:")
    print(missing)
```

**Unusual patterns/outlier Analysis**

```python
# Outlier Analysis using IQR

Q1 = df['Yield_Tonnes_Ha'].quantile(0.25)
Q3 = df['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Yield_Tonnes_Ha'] < lower_bound) |
    (df['Yield_Tonnes_Ha'] > upper_bound)
]

print("Number of Yield Outliers:", len(outliers))
print(outliers[['Farm_ID', 'State', 'District', 'Crop', 'Season',
                 'Yield_Tonnes_Ha']])
```
